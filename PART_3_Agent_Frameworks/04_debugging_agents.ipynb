## 🔍 Bug 2: Wrong Tool Selection

````python
from langchain_core.tools import tool

# BAD: Vague descriptions
@tool
def bad_search(q: str) -> str:
    """Search."""
    return "result"

# GOOD: Specific descriptions
@tool
def good_search(query: str) -> str:
    """Search the web for current information.
    
    Args:
        query: What to search for (e.g., 'weather Paris', 'AI news')
    
    Returns:
        Search results with current information
    """
    return "result"

print("❌ BAD: 'Search.' - Too vague!")
print("✅ GOOD: Full docstring with args!")
````

## 🔍 Bug 3: Tool Error Handling

````python
# BAD: No error handling
@tool
def bad_api_call(url: str) -> str:
    """API call without error handling."""
    import requests
    return requests.get(url).text  # Can fail!

# GOOD: Error handling
@tool  
def good_api_call(url: str) -> str:
    """API call with error handling."""
    import requests
    try:
        response = requests.get(url, timeout=5)
        response.raise_for_status()
        return response.text
    except requests.Timeout:
        return "Error: Request timed out"
    except requests.RequestException as e:
        return f"Error: {e}"

print("❌ BAD: No error handling - crashes agent")
print("✅ GOOD: Returns error gracefully")
````
