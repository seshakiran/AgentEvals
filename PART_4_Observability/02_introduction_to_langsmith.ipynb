## 📦 LangSmith Core Concepts

```
┌─────────────────────────────────────────────────────┐
│            LANGSMITH CONCEPTS                       │
├─────────────────────────────────────────────────────┤
│  • PROJECT: A collection of related runs       │
│  • DATASET: Inputs for testing                 │
│  • RUN: Single execution with traces         │
│  • EVALUATOR: Logic to check outputs       │
│  • TRACE: Step-by-step execution           │
└─────────────────────────────────────────────────────┘
```

## 💻 Setting Up Tracing

````python
from langsmith import traceable
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini")

@traceable(name="simple_prompt")
def simple_prompt(prompt: str) -> str:
    """A simple traced function."""
    return llm.invoke(prompt).content

# Run it
result = simple_prompt("Hello!")
print("Completed!")
print("Check LangSmith dashboard for trace!")
````

## 📊 Exploring the Dashboard

When you go to [smith.langchain.com](https://smith.langchain.com):

1. **Projects** - See all your runs
2. **Datasets** - Create test data
3. **Evaluations** - Run evals
4. **Traces** - See step-by-step execution

## 🔗 CLI Commands

````python
# Common LangSmith CLI commands:

print("""
# View runs
langsmith runs list --project "my-project"

# Create dataset
langsmith dataset create --name "my-test-data" --source_file data.jsonl

# Run evaluation
langsmith evaluate --agent my_agent --dataset "my-test-data"

# Export traces
langsmith runs export --project "my-project" --output traces.jsonl
""")
````
